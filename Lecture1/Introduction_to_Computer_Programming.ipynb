::::{seealso}
:class: dropdown

A **universal machine** (Turing machine) has three simple parts:

1. An infinitely long **tape** divided into cells, each holding a symbol
   (e.g., 0 or 1).
2. A **read/write head** that scans one cell at a time, reading the symbol
   and optionally overwriting it.
3. A **finite set of rules** (the program) that tells the machine: given the
   current symbol and internal state, what to write, which way to move (left
   or right), and what state to enter next.

Despite this simplicity, by changing the rules (the program) on the tape,
the same machine can compute *any* computable function — which is why it is
called "universal".

::::{seealso}
:class: dropdown

The three concepts are tightly intertwined:

- **Definability** asks whether a mathematical object (e.g., a set of
  numbers) can be precisely described using a formal language.
- **Computability** asks whether there exists an algorithm — a mechanical
  procedure a machine can follow — to produce or decide that object.
- **Decidability** asks whether, for any given input, there is an algorithm
  that always halts with a correct yes/no answer.

Turing showed that not every *definable* set is *computable*, and not every
computable problem is *decidable*. His proof that the Halting Problem is
undecidable means there are well-defined questions for which no algorithm
can ever give a definitive answer — establishing a fundamental limit on
what computers can do.

Unlike ENIAC, the subsequent design of the Electronic Discrete Variable Computer (EDVAC) utilized the binary number system.
````python
%%hermes
Explain the different instructions used in the different instruction set 
architectures including AMD, ARM, and RISC-V.
````

::::{seealso}
:class: dropdown

These three architectures differ fundamentally in their design philosophy:

**AMD x86/x86-64 (CISC — Complex Instruction Set Computer)**
- Variable-length instructions (1 to 15 bytes)
- Instructions can perform multi-step operations in one step, e.g., `ADD [mem], reg` reads memory, adds, and writes back simultaneously.
- Rich addressing modes allow complex memory access patterns directly inside an instruction.
- Example instructions: `MOV`, `ADD`, `PUSH`, `CALL`, `CMOVcc` (conditional move), plus SIMD extensions like `SSE`/`AVX` for parallel math on vectors.

**ARM (RISC — Reduced Instruction Set Computer)**
- Fixed-length instructions (32-bit in ARMv7/ARMv8)
- Strict load/store architecture: arithmetic only operates on registers; memory access requires explicit `LDR`/`STR` instructions.
- Fewer, simpler instructions that execute in a single cycle.
- Example instructions: `ADD`, `SUB`, `LDR`, `STR`, `B` (branch), `BL` (branch with link/call). Registers are named `X0–X30` (AArch64), plus dedicated stack pointer (`SP`) and program counter (`PC`).

**RISC-V (Open-source RISC)**
- Fixed 32-bit base instructions with optional extensions (`M` for multiply, `A` for atomics, `F/D` for floating point).
- Same load/store philosophy as ARM, with uniform register names: general-purpose registers are simply numbered `x0–x31`. Notable convention: register zero is hardwired to always return 0.
- Modular and extensible by design — you pick only the features your hardware needs.
- Example instructions: `add`, `sub`, `lw`/`sw` (load/store word), `beq`/`bne` (branch if equal/not equal), jal/jalr (jump-and-link).

**Summary:** CISC packs more work into fewer but complex instructions; RISC breaks everything into simple steps that run fast and compile cleanly.

## Agentic Coding

### What is agentic coding?

In traditional programming (3rd generation), you write **step-by-step instructions** for the computer to follow. The computer executes them literally but has no understanding of your *goal*.

In **agentic coding**, you specify a **goal** — "build a webpage that shows weather data" — and an AI agent plans, writes code, runs it, debugs errors, and iterates until the goal is met. You shift from *writing every line* to *specifying outcomes and supervising the process*.

:::{table} Comparing programming paradigms
:name: tbl:programming-generations

| Generation | Paradigm | You say... | Computer does... |
|:-----------|:---------|:-----------|:-----------------|
| 1st | Machine language | `01000001` (binary) | Executes the exact binary operation |
| 2nd | Assembly | `ADD R1, R2` | Adds two registers |
| 3rd | High-level (Python) | `for i in range(10): print(i)` | Loops and prints — follows your steps |
| **4th?** | **Agentic** | **"Create a graph showing prime number distribution"** | **Plans, codes, runs, debugs, delivers** |

:::

### How does an agent work?

An agentic coder follows a **plan-execute-reflect** loop:

1. **Plan** — Break the goal into sub-tasks
2. **Write** — Generate code for each sub-task
3. **Execute** — Run the code and observe output or errors
4. **Reflect** — If errors or incomplete results, revise and repeat

The key difference from just asking AI to "write code" is the **iteration loop**: The agent runs its own code, reads error messages, and fixes problems autonomously. You act as the *product manager*, not the *line-by-line implementer*.
