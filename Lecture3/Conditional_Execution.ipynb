## Motivation

Conditional execution means running different pieces of code based on different conditions. Why do programmers need conditional execution?

For instance, when trying to compute `a/b`, `b` may be `0` and division by `0` is invalid.

````python
%%pytutor
def multiply_or_divide(a, b):
    print("a:{}, b:{}, a*b:{}, a/b:{}".format(a, b, a * b, a / b))


multiply_or_divide(1, 2)
multiply_or_divide(1, 0)
````

Although division by 0 is invalid, multiplication remains valid but it is not printed due to the division error. Can we skip only the division but not the multiplication when `b` is equal to `0`?

**Solution 1**: Use a [*conditional
expression*](https://docs.python.org/3/reference/expressions.html#conditional-expressions)
that specifies which code block should be executed under what condition:

```python
... if ... else ...
```

````python
def multiply_or_divide(a, b):
    q = a / b if b else "undefined"
    print("a:{}, b:{}, a*b:{}, a/b:{}".format(a, b, a * b, q))


multiply_or_divide(1, 2)
multiply_or_divide(1, 0)  # multiplication is valid but not shown
````

**Solution 2**: Use a [*boolean
expression*](https://docs.python.org/3/reference/expressions.html#boolean-operations):

```python
... and ... or ...
```

````python
def multiply_or_divide(a, b):
    q = b and a / b or "undefined"
    print("a:{}, b:{}, a*b:{}, a/b:{}".format(a, b, a * b, q))


multiply_or_divide(1, 2)
multiply_or_divide(1, 0)  # multiplication is valid but not shown
````

**Solution 3**: Monitor and catch the error using a [*try
statement*](https://docs.python.org/3/reference/compound_stmts.html#try):

````python
def multiply_or_divide(a, b):
    try:
        q = a / b
    except ZeroDivisionError:
        q = "undefined"
    print("a:{}, b:{}, a*b:{}, a/b:{}".format(a, b, a * b, q))


multiply_or_divide(1, 2)
multiply_or_divide(1, 0)  # multiplication is valid but not shown
````

In this notebook, we will introduce the first two solutions but the last one
can be a better solution for more complicated scenarios.

The **Entscheidungsproblem** (German: "decision problem") was posed by **David Hilbert** in 1928, formally stated in Hilbert and Ackermann's book *Principles of Mathematical Logic*. It asked:

> *Can every well-formed mathematical statement be proven true or false algorithmically?*

In other words: is there a general mechanical procedure that, given any mathematical statement, can always decide whether it is true or false?

**Turing proved the answer is NO** — in his 1936 paper, he showed no such algorithm can exist. This is one of the foundational results of computer science: there are fundamental limits to what any computational process can decide.

The connection runs through the chain of ideas in the notes:

1. **George Boole (1815–1864)** created **Boolean algebra** — the system where logic is arithmetic. Truth values are just numbers (TRUE = 1, FALSE = 0), and logical operations (AND, OR, NOT) are arithmetic operations. His key insight: *logic can be computed*.

2. **Hilbert's programme** (early 20th century) asked whether this algebraic approach could be made complete: can *all* true mathematical statements be proven? And is there an algorithm to *decide* truth? — the Entscheidungsproblem.

3. **Turing's negative answer** to the Entscheidungsproblem showed that despite Boole's beautiful reduction of logic to algebra, there are mathematical truths that **no algorithm** — no matter how clever — can decide.

The irony, as the notes put it: *"Logic is arithmetic. Arithmetic has no general answer machine."* Boole showed logic *could* be computed, but Turing showed computation has **inherent limits** — some questions are fundamentally undecidable.

### Comparing Floating Point Numbers

`==` can also be used to compare floating point numbers:

````python
x = 10
y = (x ** (1 / 3)) ** 3
x == y
````

::::{caution} Why False? Shouldn't $(x^{\frac13})^3=x$?
:class: dropdown

Floating point numbers have finite precisions and so the computation $x^{\frac13}$ may not be exact.

::::

In practice, a small error in the calculation can be tolerated. Instead of checking equality of floating point numbers, we can check closeness up to a specified tolerance:

````python
abs_tol = 1e-9
y - abs_tol <= x <= y + abs_tol
````

`abs_tol`, often denoted as $\delta_{\text{abs}}$, is a positive number called the *absolute tolerance*.

**Why call it absolute tolerance?**

Note that the test remains unchanged if we swap `x` and `y`:

````python
abs_tol = 1e-9
x - abs_tol <= y <= x + abs_tol
````

Using the absolute function `abs`, we can also rewrite the comparison as follows:

````python
abs_tol = 1e-9
abs(x - y) <= abs_tol
````

::::{prf:definition} Absolute tolerance

The equality of $x$ and $y$ within an absolute tolerance of $\delta_{\text{abs}}$ means

$$x=y\pm \delta_{\text{abs}},$$

or more explicitly,

$$\lvert x-y\rvert \leq \delta_{\text{abs}}.$$

In other words, the comparison does not tolerate an absolute difference between $x$ and $y$ larger than $\delta_{\text{abs}}$.

::::

**Is an absolute tolerance of `1e-9` good enough?**

The same absolute tolerance fails if we set `x = 1e10` instead of `10`.

````python
x = 1e10
y = (x ** (1 / 3)) ** 3

abs_tol = 1e-9
abs(x - y) <= abs_tol
````

::::{caution} Why absolute tolerance fails?
:class: dropdown

In floating-point representation, the [radix point](https://simple.wikipedia.org/wiki/Radix_point) *floats* across different scales according to the exponent. The precision error in the mantissa is therefore relative to the scale of the number.

::::

We should instead check whether $x$ is within a certain percentage of $y$:

````python
rel_tol = 1e-9
y * (1 - rel_tol) <= x <= y * (1 + rel_tol)
````

Note that the above test is not symmetric between $x$ and $y$, i.e., $x$ is relatively close to $y$ does not necessarily mean $y$ is relatively close to $x$.

````python
x = 1
y = 2
rel_tol = 0.5
y * (1 - rel_tol) <= x <= y * (1 + rel_tol), x * (1 - rel_tol) <= y <= x * (1 + rel_tol)
````

To make the comparison symmetric, we can use the absolute difference as follows:

::::{prf:definition} Relative tolerance

The equality of $x$ and $y$ is within a *relative tolerance* of $\delta_{\text{rel}}$ if

$$x=y \pm \delta_{\text{rel}} \max\Set{\lvert x\rvert,\lvert y\rvert},$$

or equivalently,

$$\lvert x-y\rvert \leq  \delta_{\text{rel}} \max\Set{\lvert x\rvert,\lvert y\rvert}.$$ (eq:rel_tol)

::::

::::{note} Why use $\max\Set{\lvert x\rvert,\lvert y\rvert}$ but not $\min\Set{\lvert x\rvert,\lvert y\rvert}$?
:class: dropdown

This is to make the test more tolerant, i.e.,

$$
\begin{align}
x&=y (1 \pm \delta_{\text{rel}}) && \text{ or } & y&=x (1 \pm \delta_{\text{rel}})
\end{align}
$$

instead of requiring both.

::::{caution} When does relative tolerance fail?
:class: dropdown

If $x$ or $y$ is very close to 0, the relative tolerance test becomes ineffective, and so absolute tolerance is a more meaningful test.
In particular, with $y=0$, [](#eq:rel_tol) becomes 

$$
\begin{align}
\lvert x-0\rvert&\leq \delta_{\text{rel}} \max\Set{\lvert x\rvert, 0}\\
\lvert x\rvert&\leq \delta_{\text{rel}} \lvert x\rvert\\
1 &\leq \delta_{\text{rel}} \quad\text{or}\quad x=0.
\end{align}
$$

In the nontrivial case $x\neq 0$, the test $\delta_{\text{rel}}\geq 1$ is independent of the value of $x$, and therefore cannot check whether $x$ is close to $0$. In particular, since $\delta_{\text{rel}}<1$ normally, and the test would always fail unless $x=0$ exactly, which defeats the purpose of allowing a tolerance.

::::{prf:definition} absolute and relative tolerances combined

Comparing floating-point numbers with `==` is dangerous because floats are stored as binary approximations in IEEE 754 format, so many decimal values (like `0.1`) cannot be represented exactly. Arithmetic operations accumulate tiny rounding errors, so `0.1 + 0.2` doesn't equal `0.3` bit-for-bit — it's `0.30000000000000004`. In real-world engineering, this means a control system's threshold check (`if temp == 100.0`) can silently fail, a financial calculation can report mismatched balances after repeated additions, and physics simulations can produce divergent results depending on the order of operations. The fix is to compare within a tolerance (`abs(a - b) < epsilon`), not with exact equality.

## Conditional Constructs

To illustrate how Python can carry out conditional execution, we will consider writing a program that sorts values in *ascending* order.

### If-Then Construct

**How to sort two values?**

Given two values are stored as `x` and `y`, we want to 
- `print(x,y)` if `x <= y`, and
- `print(y,x)` if `y < x`.

Such a program flow is often represented by a flowchart like the following:

````python
sort_two_values_fc1
````

**How to read the flowchart?**

A flowchart uses arrows to connects a set of [annotated blocks][bb]. The rules were first specified by ANSI and later adopted in [ISO 5807][iso].

[bb]: https://en.wikipedia.org/wiki/Flowchart#Building_blocks
[iso]: https://webstore.ansi.org/Standards/ISO/ISO58071985

**Why to use a program flowchart?**

A program flowchart is a powerful way of describing an algorithm quickly. Unlike a text-based programming language:
- The rules governing the program flow can be shown explicitly by arrows.
- The annotated graphical blocks can convey the meaning faster using visual clues.

**How to implement the flowchart in Python?**

It is often useful to delay detailed implementations until we have written an overall skeleton. To leave a block empty, Python uses the keyword [`pass`](https://docs.python.org/3/tutorial/controlflow.html#pass-statements).

````python
# write a code skeleton
def sort_two_values(x, y):
    pass
    # print the smaller value first followed by the larger one


sort_two_values(1, 0)
sort_two_values(1, 2)
````

Python provides the [`if` statement](https://docs.python.org/3/reference/compound_stmts.html#the-if-statement) to implement the [*control flow*](https://en.wikipedia.org/wiki/Control_flow) specified by the diamond boxes in the flowchart.

````python
def sort_two_values(x, y):
    if x <= y:
        pass
        # print x before y
    if y < x: pass  # print y before x


sort_two_values(1, 0)
sort_two_values(1, 2)
````

To complete the implementations specified by the parallelogram boxes in the flow chart, we fill in the bodies/suites of the `if` statements as follows:

````python
def sort_two_values(x, y):
    if x <= y:
        print(x, y)
    if y < x: print(y, x)


@interact(x="1", y="0")
def sort_two_values_app(x, y):
    print("Values in ascending order:")
    sort_two_values(eval(x), eval(y))
````

### Nested Conditionals

Now, consider a slightly more challenging problem of sorting three values instead of two. A feasible algorithm is as follows:

````python
sort_three_values_fc
````

To implement flowchart, we can use *nested conditional constructs*, where one conditional statement is placed within another, allowing for multiple layers of condition checks:

````python
def sort_three_values(x, y, z):
    if x <= y <= z:
        print(x, y, z)
    else:
        if x <= z <= y:
            print(x, z, y)
        else:
            if y <= x <= z:
                print(y, x, z)
            else:
                if y <= z <= x:
                    print(y, z, x)
                else:
                    if z <= x <= y:
                        print(z, x, y)
                    else:
                        print(z, y, x)


def test_sort_three_values():
    sort_three_values(0, 1, 2)
    sort_three_values(0, 2, 1)
    sort_three_values(1, 0, 2)
    sort_three_values(1, 2, 0)
    sort_three_values(2, 0, 1)
    sort_three_values(2, 1, 0)


test_sort_three_values()
````

Imagine what would happen if we have to sort many values. The program will not only be long, but also fat due to the indentation. To avoid an excessively long line due to the indentation, Python provides the `elif` keyword that combines `else` and `if`.

````python
def sort_three_values(x, y, z):
    if x <= y <= z:
        print(x, y, z)
    elif x <= z <= y:
        print(x, z, y)
    elif y <= x <= z:
        print(y, x, z)
    elif y <= z <= x:
        print(y, z, x)
    elif z <= x <= y:
        print(z, x, y)
    else:
        print(z, y, x)


test_sort_three_values()
````

::::{seealso}
:class: dropdown

The minimum number of comparisons needed to sort *n* numbers is Ω(*n* log *n*) in the worst case, because comparison-based sorting can be modeled as a decision tree with *n!* leaf nodes (one per possible permutation), and a binary tree with *L* leaves has height at least ⌈log₂ *L*⌉. This is fundamentally different from the maximum number of inversions, which is *n*(*n*−1)/2 — that counts how many pairs are out of order in the worst case (a reversed list), not how many *questions* a smart algorithm must ask. One is a **lower bound on information needed** (log of permutations), the other is an **upper bound on pairwise disorder** — they measure completely different things, which is why insertion sort can take O(*n*²) swaps in the worst case while mergesort always runs in O(*n* log *n*).

## Boolean Operations

Since chained comparisons are non-associative, it follows a different evaluation rule than arithmetic operators.

E.g., `1 <= 2 < 3 != 4` is equivalent to:

````python
1 <= 2 and 2 < 3 and 3 != 4
````

The above is called a *compound boolean expression*, which is formed using the
*boolean/logical operator* `and`.

**Why use boolean operators?**

What if we want to check whether a number is either $< 0$ or $\geq 100$?

Turns out this is mission impossible because `and` alone is not [functionally
complete](https://en.wikipedia.org/wiki/Functional_completeness),  i.e., it is
not enough to give all possible boolean functions. We can add `or` and `not`
operators to make it functionally complete.

::::

````python
%%hermes
Explain in a paragraph why the set of operators including only "and" is not 
functionally complete.
````

::::{seealso}
:class: dropdown

The set {AND} is not functionally complete because AND is **truth-preserving**: whenever all inputs are True, the output is always True, no matter how you combine them. This means you can never produce a NOT operation — if you feed only True values into any expression built from AND alone, you'll never get False. Since there are 2^(2ⁿ) possible Boolean functions on *n* variables (including functions like NOT, NOR, and NAND that *do* output False when all inputs are True), and AND can only express a tiny subset of them, it alone cannot represent every possible Boolean function.

::::

````python
%%hermes
Explain in a paragraph why the set of operators "and", "or", and "not" is 
functionally complete.
````

::::{seealso}
:class: dropdown

The set {AND, OR, NOT} is functionally complete because any Boolean function can be expressed as a **disjunctive normal form (DNF)** — a combination of ANDs, ORs, and NOTs. Specifically, every truth table row where the output is True can be captured by one AND-term (the conjunction of all inputs, each either as-is or negated), and OR-ing those terms together reconstructs the entire function. Since this construction works for *any* truth table, {AND, OR, NOT} can express all 2^(2ⁿ) possible Boolean functions on *n* variables.

::::

### Short-circuit `or`

Consider evaluating `True or False and True` in a verbose manner:

````python
verbose("A", verbose(1, True) or verbose(2, False) and verbose(3, True))
````

::::{note}

When evaluating `True or False and True`,

- even though `or` has lower precedence than `and`, the short-circuit evaluation for `or` is performed first.
- The `and` operation is skipped entirely.

::::

**Why the second and third expressions are not evaluated?**

Because `True or ...` must be `True` (why?) so Python does not look further. From the [documentation](https://docs.python.org/3/reference/expressions.html?highlight=precedence#boolean-operations):

::::{card}
:header: Short-circuit evaluation of `or`

The expression `x or y` 
1. first evaluates `x`;  
2. if `x` is true, its value is returned;  
3. otherwise, `y` is evaluated and the resulting value is returned.

::::

Put it another way, `(x or y)` translates to the following

:::{code-block} python
:name: code:or
:caption: Translation of `(x or y)`

_ if (_ := x) else y

:::

except for the extra variable `_`. For example, `True or False and True` translates to `_ if (_ := True) else False and True`:

````python
verbose("A", (_ if (_ := verbose(1, True)) else verbose(2, False) and verbose(3, True)))
````

### Short-circuit `and`

Now, consider evaluating `False or False and True` in a verbose manner:

````python
False or False and True
````

````python
verbose("B", verbose(4, False) or verbose(5, False) and verbose(6, True))
````

**Why expression 6 is not evaluated?**

`False or False and ...` must be `False` so Python does not look further.

::::{card}
:header: Short-circuit evaluation of `and`

The expression `x and y` first evaluates `x`;  
if `x` is false, its value is returned;  
otherwise, `y` is evaluated and the resulting value is returned.
::::

Put it another way, `(x and y)` translate to the following

:::{code-block} python
:name: code:and
:caption: Translation of `(x and y)`

_ if not (_ := x) else y

:::


except for the extra variable `_`. For example, `False or False and True` translates to `False or (_ if not (_ := False) else True)`:

````python
verbose(
    "B", verbose(4, False) or (_ if not (_ := verbose(5, False)) else verbose(6, True))
)
````

In Python, `True` and `False` are subclasses of `int` where `True == 1` and `False == 0`, so `0 == False` is `True` and `1 == True` is also `True`, making the whole expression evaluate to `True`.
````python
%%hermes
Give a real-world example where short-circuit evaluation is used for safety
checks before performing expensive or dangerous operations.
````

::::{seealso}
:class: dropdown

A classic example is checking if a pointer/reference exists before accessing it:

```python
if user is not None and user.has_admin_privileges():
    delete_database()
```

Here, `user is not None` is evaluated first. If `user` is `None`, the second part (`user.has_admin_privileges()`) is **never executed** — preventing a `NoneType` error that would crash the program (or worse, skip the authorization check entirely). Without short-circuit evaluation, the code would try to call a method on `None`, which would either crash or bypass the safety guard, potentially allowing unauthorized access to `delete_database()`.
